```python
import pandas as pd
data ={"popcorn" : [1,1,0,0,1,1,0],
"soda" : [1,0,1,1,1,0,0],
"age" : [7,12,18,35,38,50,83],
"Movie" : [0,0,1,1,1,0,0]
}

df = pd.DataFrame(data)
df
```

```python
x = df[["popcorn","soda","age"]] #feature
y = df["Movie"]#Prediction Target
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model = DecisionTreeClassifier(criterion="gini")
model.fit(x,y)
```

```python
pred = [[0,0,25]]
prediction = model.predict(pred)
prediction
```

```python
from sklearn.tree import plot_tree
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,4))
plot_tree(model,feature_names=["popcorn","soda","age"],class_names=["yes","no"],filled = True)
plt.show
```
